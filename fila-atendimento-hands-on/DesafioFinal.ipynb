```python
import names
import random
import heapq

class Cliente:
    def __init__(self, nome, senha, prioridade):
        self.nome = nome
        self.senha = senha
        self.prioridade = prioridade

    def __str__(self):
        return f"Nome: {self.nome} | Senha: {self.senha} | Prioridade: {self.prioridade}"


# Criação de uma fila base para as operações.
clientes = []

for i in range(1, 21):
    nome = names.get_full_name()
    prioridade = random.randint(1, 3)
    senha = f"{i:03d}"


    cliente = Cliente(nome, senha, prioridade)
    clientes.append(cliente)

print("=== 20 CLIENTES GERADOS ===")

for cliente in clientes:
    print(cliente)

# ----------------------------------------------

# implementação das estruturas de Fila Clássica e Circular:


# Fila Clássica:

class Fila:
  def __init__(self):
    self.dados = []

# função para tornar o output em string légivel
  def __str__(self):
    return str(self.dados)

# enqueue
  def enqueue(self, elemento):
    self.dados.append(elemento)

# dequeue
  def dequeue(self):
    if not self.empty():
      return self.dados.pop(0)
    return None

# head
  def head(self):
    if not self.empty():
      return self.dados[0]
    return None

# size
  def size(self):
    return len(self.dados)

# empty
  def empty(self):
    return len(self.dados) == 0


# Fila Circular:

class FilaCircular:
  def __init__(self, capacidade):
    self.vetor = [None] * capacidade
    self.capacidade = capacidade
    self.front = -1
    self.rear = -1

  def __str__(self):
    # Para uma representação mais clara da fila circular
    if self.is_empty():
        return "Fila Circular: []"

    elementos = []
    if self.rear >= self.front:
        for i in range(self.front, self.rear + 1):
            elementos.append(str(self.vetor[i].senha) if self.vetor[i] else "None")
    else:
        for i in range(self.front, self.capacidade):
            elementos.append(str(self.vetor[i].senha) if self.vetor[i] else "None")
        for i in range(0, self.rear + 1):
            elementos.append(str(self.vetor[i].senha) if self.vetor[i] else "None")

    return f"Fila Circular: [{', '.join(elementos)}] (Front: {self.front}, Rear: {self.rear}, Capacidade: {self.capacidade})"

# empty
  def is_empty(self):
    return self.front == -1

# full
  def is_full(self):
    return (self.rear + 1) % self.capacidade == self.front

# enqueue
  def enqueue(self, valor):
    if self.is_full():
      print("Fila cheia!")
      return
    if self.is_empty():
      self.front = 0

    self.rear = (self.rear + 1) % self.capacidade
    self.vetor[self.rear] = valor

    print(f"{valor.senha} entrou na posição {self.rear}.\n") #

# dequeue
  def dequeue(self):
    if self.is_empty():
      print("Fila vazia!")
      return None

    valor = self.vetor[self.front]
    self.vetor[self.front] = None

    if self.front == self.rear:
      self.front = self.rear = -1
    else:
      self.front = (self.front + 1) % self.capacidade

    return valor

# head
  def head(self):
    if self.is_empty():
      return None
    return self.vetor[self.front]

# tail
  def tail(self):
    if self.is_empty():
        return None
    return self.vetor[self.rear]

# size
  def size(self):
    if self.is_empty():
      return 0
    if self.rear >= self.front:
      return self.rear - self.front + 1
    else:
      return self.capacidade - self.front + self.rear + 1

# ----------------------------------------------

# Implementação básica da Fila de Prioridade usando heapq
class FilaPrioridade:
    def __init__(self):
        self._queue = []
        self._index = 0

    def enqueue(self, cliente):
        heapq.heappush(self._queue, (-cliente.prioridade, self._index, cliente))
        self._index += 1

    def dequeue(self):
        if not self.empty():
            return heapq.heappop(self._queue)[2]  # Retorna o objeto cliente
        return None

    def empty(self):
        return len(self._queue) == 0


# FILA CLÁSSICA

fila = Fila()

for cliente in clientes:
    fila.enqueue(cliente)

ordem_classica = []

print("\n=== ATENDIMENTO - FILA CLÁSSICA ===")

while not fila.empty():
    cliente = fila.dequeue()
    ordem_classica.append(cliente)
    print(cliente)

# FILA CIRCULAR

fila_circular = FilaCircular(5)

ordem_circular = []

print("\n=== ATENDIMENTO FILA CIRCULAR - CAPACIDADE 5 ===")

for cliente in clientes:

    fila_circular.enqueue(cliente)

    print("\nInserido:", cliente.senha)
    print(fila_circular) #

    if fila_circular.size() == 5:
        atendido = fila_circular.dequeue()
        ordem_circular.append(atendido)

        print("Atendido:", atendido.senha)
        print(fila_circular) #

print("\n=== FINALIZANDO ATENDIMENTO DA FILA CIRCULAR ===")

while not fila_circular.is_empty(): #
    atendido = fila_circular.dequeue()
    ordem_circular.append(atendido)

    print("\nAtendido:", atendido.senha)
    print(fila_circular) #

# FILA DE PRIORIDADE

fila_prioridade = FilaPrioridade()

for cliente in clientes:
    fila_prioridade.enqueue(cliente)

ordem_prioridade = []

print("\n=== ATENDIMENTO - FILA DE PRIORIDADE ===")

while not fila_prioridade.empty():
    atendido = fila_prioridade.dequeue()
    ordem_prioridade.append(atendido)
    print("Atendido:", atendido)


print("\n=== COMPARAÇÃO FINAL ===")

print("\nFila clássica:")
for cliente in ordem_classica:
    print(cliente.senha, end=" -> ")

print("\n\nFila circular:")
for cliente in ordem_circular:
    print(cliente.senha, end=" -> ")

print("\n\nFila de prioridade:")
for cliente in ordem_prioridade:
    print(cliente.senha, end=" -> ")

print("\n")
```
