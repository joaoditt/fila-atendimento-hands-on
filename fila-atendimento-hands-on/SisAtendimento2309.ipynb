 # Parte 1 - Fila Clássica

```python
class Fila:
  def __init__(self):
    self.dados = []

# função para tornar o output em string légivel
  def __str__(self):
    return str(self.dados)

# enqueue
  def enqueue(self, elemento):
    self.dados.append(elemento)

# dequeue
  def dequeue(self):
    if not self.empty():
      return self.dados.pop(0)
    return None

# head
  def head(self):
    if not self.empty():
      return self.dados[0]
    return None

# size
  def size(self):
    return len(self.dados)

# empty
  def empty(self):
    return len(self.dados) == 0


# implementação da fila com cadastro de 10 clientes:
fila = Fila()

# (prioridade, senha/ordem, nome)

fila.enqueue((2, "1", "Joana"))
fila.enqueue((1, "2", "Luiz"))
fila.enqueue((2, "3", "Arnaldo"))
fila.enqueue((3, "4", "Kaíque"))
fila.enqueue((2, "5", "Gabriela"))
fila.enqueue((1, "6", "Bruna"))
fila.enqueue((3, "7", "Torres"))
fila.enqueue((3, "8", "Paula"))
fila.enqueue((1, "9", "Charles"))
fila.enqueue((2, "10", "André"))

print("Fila inicial:", fila)

for i in range(11):

  print("\nCliente atendido:", fila.head())
  print("Tamanho da fila:", fila.size())
  print("Fila atual:", fila)

  if (fila.empty()):
    print("\nFila vazia.")

  else:
    fila.dequeue()
```

# Parte 2 - Fila Circular

```python
class FilaCircular:
  def __init__(self, capacidade):
    self.vetor = [None] * capacidade
    self.capacidade = capacidade
    self.front = -1
    self.rear = -1

# função para tornar o output em string légivel
  def __str__(self):
    return str(self.vetor)

# empty
  def is_empty(self):
    return self.front == -1

# full
  def is_full(self):
    return (self.rear + 1) % self.capacidade == self.front

# enqueue
  def enqueue(self, valor):
    if self.is_full():
      print("Fila cheia!")
      return
    if self.is_empty():
      self.front = 0

    self.rear = (self.rear + 1) % self.capacidade
    self.vetor[self.rear] = valor

    print(f"{valor[2]} entrou na posição {self.rear}.\n")

# dequeue
  def dequeue(self):
    if self.is_empty():
      print("Fila vazia!")
      return None

    valor = self.vetor[self.front]
    self.vetor[self.front] = None

    if self.front == self.rear:
      self.front = self.rear = -1
    else:
      self.front = (self.front + 1) % self.capacidade

    return valor

# head
  def head(self):
    if self.is_empty():
      return None
    return self.vetor[self.front]

# tail
  def tail(self):
    if self.is_empty():
        return None
    return self.vetor[self.rear]

# size
  def size(self):
    if self.is_empty():
      return 0
    if self.rear >= self.front:
      return self.rear - self.front + 1
    else:
      return self.capacidade - self.front + self.rear + 1


# Implementação da Fila Circular e 5 Clientes Cadastrados:

fila = FilaCircular(5)

# (prioridade, senha/ordem de chegada, nome)

fila.enqueue((2, "Senha: 1", "Joana"))
fila.enqueue((1, "Senha: 2", "Luiz"))
fila.enqueue((2, "Senha: 3", "Arnaldo"))
fila.enqueue((3, "Senha: 4", "Kaíque"))
fila.enqueue((2, "Senha: 5", "Gabriela"))

print("Fila inicial:", fila)

print("\nÍndice inicial (front):", fila.front)
print("Cliente no front:", fila.head())

print("\nÍndice final (rear):", fila.rear)
print("Cliente no rear:", fila.tail())

print("\nTamanho da fila:", fila.size())

# Clientes com indíce/posição '0' e '1' são atendidos:
print("\nClientes Joana e Luiz atendidos.\n")
fila.dequeue()
fila.dequeue()

print("\nFila atual:", fila)

print("\nÍndice inicial (front):", fila.front)
print("Cliente no front:", fila.head())

print("\nÍndice final (rear):", fila.rear)
print("Cliente no rear:", fila.tail())

print("\nTamanho da fila:", fila.size())

# Novo Cliente entra no indíce '0':
print("\nNovo cliente entra na fila.\n")
fila.enqueue((1, "Senha: 6", "Bruna"))

print("\nFila atual:", fila)

print("\nÍndice inicial (front):", fila.front)
print("Cliente no front:", fila.head())

print("\nÍndice final (rear):", fila.rear)
print("Cliente no rear:", fila.tail())

print("\nTamanho da fila:", fila.size())

```

```python


# bibliotecas para geração de nomes e prioridade aleatórias.
import names
import random

clientes = []

for contador in range(1, 21):
    prioridade = random.randint(1, 3)
    nome = names.get_full_name()

    clientes.append((prioridade, contador, nome))

print("Prioridades: \n1 - Emergência\n2 - Prioritário\n3 - Atendimento Normal\n--------//--------")

print("\nLista de Clientes:\n")
for cliente in clientes:
    print(cliente)


import heapq # biblioteca necessária para a Fila de Prioridade.

heap = []

# inserção dos clientes na fila heap.
for cliente in clientes:
    heapq.heappush(heap, cliente)

print("\nFila de Prioridade:\n")

while heap:
    cliente = heapq.heappop(heap)
    print(cliente)
```
